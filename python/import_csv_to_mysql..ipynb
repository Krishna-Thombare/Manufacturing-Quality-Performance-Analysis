### 1. Make connection with MySQL Workbench

```python
# Establish a connection with MySQL

from sqlalchemy import create_engine
from sqlalchemy.engine import URL
from dotenv import load_dotenv
import os

load_dotenv()

connection_url = URL.create(
    "mysql+pymysql",
    username="root",
    password=os.getenv("PASSWORD"),
    host="localhost",
    port=3306,
    database=os.getenv("DATABASE")
)

engine = create_engine(connection_url)
```

### 2. Create columns with proper Data Types

```python
from sqlalchemy.types import (
    String, Date, Time, Integer, Numeric, Boolean
)

# Create MySQL data types
column_types = {
    "inspection_id": String(30),
    "inspection_date": Date(),
    "inspection_time": Time(),
    "plant_location": String(100),
    "production_line": String(30),
    "machine_id": String(30),
    "shift": String(20),
    "inspection_type": String(10),
    "inspector_id": String(30),
    "product_id": String(30),
    "product_name": String(100),
    "product_category": String(50),
    "supplier_name": String(100),
    "batch_id": String(40),
    "sample_size": Integer(),
    "units_defective": Integer(),
    "defect_type": String(100),
    "defect_severity": String(20),
    "spec_dimension_mm": Numeric(12, 3),
    "tolerance_mm": Numeric(12, 3),
    "measured_dimension_mm": Numeric(12, 3),
    "measured_weight_g": Numeric(12, 3),
    "process_temp_c": Numeric(12, 2),
    "inspection_duration_min": Numeric(12, 2),
    "inspection_result": String(20),
    "root_cause_category": String(50),
    "rework_cost_inr": Numeric(12, 2),
    "scrap_cost_inr": Numeric(12, 2),
    "downstream_complaint": String(10),
    "duplicate_id_conflict": Boolean()
}
```

### 3. Load the Cleaned Dataset

```python
import pandas as pd

df_work = pd.read_csv(
    "../data/product_quality_control_cleaned.csv",
    parse_dates=["inspection_date"],
    keep_default_na=False, # keeps the real value "None"
    na_values=[""]   # only truly empty cells become NULL
)
```

```python
print(df_work.shape)
print(df_work.dtypes)
```

```python
print(df_work["defect_type"].eq("None").sum())   
print(df_work["defect_type"].eq("Unknown").sum())
```

### 4. Import CSV to MySQL Workbench

```python
df_work.to_sql(
    name="quality_inspections",
    con=engine,
    if_exists="replace",
    index=False,
    dtype=column_types,
    chunksize=1000
)

print("Data imported successfully!")
```
